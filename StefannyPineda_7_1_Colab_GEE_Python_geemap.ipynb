# Bloque 1. Computación en la nube aplicada a Big Data geoespacial

## 1.1 Modelos de servicio

| Modelo | Qué entrega el proveedor | Ejemplo geoespacial |
|---|---|---|
| **IaaS** (infraestructura) | Máquinas virtuales, discos, red | Un clúster Spark propio en AWS EC2 / GCP Compute; PostGIS en una VM |
| **PaaS** (plataforma) | Servicios gestionados: bases de datos, colas, procesamiento | BigQuery GIS, Dataproc (Spark gestionado), Azure Batch |
| **SaaS** (software) | Aplicación lista para usar vía API o web | **Google Earth Engine**, Planetary Computer, ArcGIS Online, Sentinel Hub |

## 1.2 ¿Por qué la nube para datos geoespaciales masivos?

* **Los datos ya están allí.** Sentinel, Landsat, MODIS, DEM, clima: petabytes alojados por Google, AWS, Microsoft y ESA.
  Mover 1 PB es inviable; mover el algoritmo cuesta nada.
* **Escalado elástico:** miles de núcleos durante minutos, pago por uso (o gratis para investigación/educación).
* **Almacenamiento de objetos + formatos nativos de nube** (Clase 5): COG, GeoParquet, Zarr.
* **Servicios gestionados**: nadie administra servidores; el analista se concentra en el modelo.

## 1.3 El modelo de Google Earth Engine

Earth Engine (EE) es una plataforma **SaaS** de análisis planetario: un catálogo de > 90 PB de imágenes y tablas, un
motor de procesamiento distribuido y una API (JavaScript y **Python**). Conceptos clave:

| Concepto | Descripción |
|---|---|
| **Evaluación perezosa** | Las instrucciones construyen un *grafo* en el cliente; nada se calcula hasta que se pide un resultado (`getInfo`, un mapa, una exportación). Igual que Spark. |
| **Cómputo bajo demanda por teselas** | Al mostrar un mapa, EE calcula solo las teselas visibles a la escala visible. Por eso "todo es instantáneo" en el visor. |
| **Escala y proyección** | EE remuestrea automáticamente a la escala solicitada (`scale` en metros). Pedir estadísticas a 10 m sobre Colombia entera es una petición enorme: elija escalas razonables. |
| **Cliente vs. servidor** | Los objetos `ee.*` viven en el servidor; `getInfo()` trae un objeto al cliente (lento, limitado). Evite ciclos `for` de Python sobre objetos de EE: use `map()` del servidor. |
| **Límites** | `getInfo` ≤ 5.000 elementos por llamada en tablas y tiempo de espera de 5 min; para más, use `Export`. |
| **Assets** | Sus propias tablas e imágenes subidas a EE (hasta 250 GB gratis). |

**Comparación con Spark (Clase 5):** ambos construyen un plan perezoso y lo ejecutan distribuido; Spark procesa
*sus* tablas en *su* clúster; EE procesa el catálogo de Google en la infraestructura de Google, con operaciones
ráster especializadas.

```python
import importlib, subprocess, sys
for paquete, modulo in [("earthengine-api", "ee"), ("geemap", "geemap")]:
    if importlib.util.find_spec(modulo) is None:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", paquete], check=False)
import os, time, json, warnings
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import ee, geemap
warnings.filterwarnings("ignore")

# Autenticación. Reemplace por el ID de su proyecto de Google Cloud registrado en Earth Engine.
PROYECTO_GEE = "accesoparques"
try:
    ee.Initialize(project=PROYECTO_GEE)
except Exception:
    ee.Authenticate()                     # abre un enlace para iniciar sesión con la cuenta de Google
    ee.Initialize(project=PROYECTO_GEE)
print("Earth Engine inicializado en", PROYECTO_GEE, "| geemap", geemap.__version__)

def fc_a_df(fc):
    # FeatureCollection → DataFrame (≤ 5.000 elementos por llamada). Añade lon/lat si hay geometría de punto.
    feats = fc.getInfo()["features"]
    df = pd.DataFrame([f["properties"] for f in feats])
    if feats and feats[0].get("geometry") and feats[0]["geometry"]["type"] == "Point":
        df["lon"] = [f["geometry"]["coordinates"][0] for f in feats]; df["lat"] = [f["geometry"]["coordinates"][1] for f in feats]
    return df
```

# Bloque 3. Objetos básicos y evaluación perezosa

Todo lo que empieza por `ee.` es un **objeto del servidor**: una descripción de un cálculo, no un dato. Observe la
diferencia entre crear un objeto (instantáneo) y pedir su valor con `getInfo()` (viaja al servidor).

```python
# Geometrías y features
punto = ee.Geometry.Point([-73.1227, 7.1193])                       # Bucaramanga (lon, lat)
aoi = ee.Geometry.Rectangle([-73.25, 7.00, -73.00, 7.25])            # caja de ~27 × 27 km
municipio = ee.Feature(aoi, {"nombre": "Bucaramanga AM", "fuente": "caja manual"})

# Una imagen: el DEM SRTM (30 m)
srtm = ee.Image("USGS/SRTMGL1_003")
t = time.time(); print("Objeto creado (sin cómputo):", type(srtm), f"{time.time()-t:.4f} s")

t = time.time(); info = srtm.getInfo(); print(f"getInfo() de los metadatos: {time.time()-t:.2f} s | bandas:", [b["id"] for b in info["bands"]])

# Operaciones ráster en el servidor: pendiente y orientación derivadas del DEM
pendiente = ee.Terrain.slope(srtm)
t = time.time()
stats = srtm.addBands(pendiente).reduceRegion(ee.Reducer.mean().combine(ee.Reducer.minMax(), sharedInputs=True),
                                              geometry=aoi, scale=90, maxPixels=1e9).getInfo()
print(f"Estadísticas en el AOI ({time.time()-t:.1f} s):", {k: round(v, 1) for k, v in stats.items()})
```

```python
# Colecciones: Sentinel-2 L2A (reflectancia de superficie, armonizada) sobre el AOI
s2 = (ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
        .filterBounds(aoi)
        .filterDate("2024-01-01", "2024-03-31")
        .filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 20)))
print("Escenas:", s2.size().getInfo())
tabla = pd.DataFrame(s2.aggregate_array("system:index").getInfo(), columns=["id"])
tabla["nubes_%"] = s2.aggregate_array("CLOUDY_PIXEL_PERCENTAGE").getInfo()
tabla.head(8)
```

> **Regla de oro:** `aggregate_array`, `reduceRegion`, `size` traen resultados pequeños. Nunca haga `getInfo()` de
> una imagen completa ni itere en Python sobre `s2.toList(n)`: use `map` en el servidor.

# Bloque 4. Enmascarar nubes, componer y calcular índices

Un **compuesto** reduce una colección a una imagen (mediana por píxel) eliminando nubes píxel a píxel con la banda de
clasificación de escena `SCL` (o la máscara `MSK_CLDPRB`). Luego calculamos índices y los visualizamos con `geemap`.

```python
def enmascarar_nubes(img):
    # SCL: 3 sombra de nube, 8 nube media, 9 nube alta, 10 cirro, 11 nieve → se descartan
    scl = img.select("SCL")
    mascara = scl.neq(3).And(scl.neq(8)).And(scl.neq(9)).And(scl.neq(10)).And(scl.neq(11))
    return img.updateMask(mascara).divide(10000).copyProperties(img, ["system:time_start"])

def agregar_indices(img):
    ndvi = img.normalizedDifference(["B8", "B4"]).rename("NDVI")
    ndwi = img.normalizedDifference(["B3", "B8"]).rename("NDWI")
    ndbi = img.normalizedDifference(["B11", "B8"]).rename("NDBI")
    return img.addBands([ndvi, ndwi, ndbi])

compuesto = agregar_indices(s2.map(enmascarar_nubes).median()).clip(aoi)
print("Bandas del compuesto:", compuesto.bandNames().getInfo())

Map = geemap.Map(center=[7.12, -73.12], zoom=11, basemap="HYBRID")

# Mapas base adicionales, seleccionables desde el control de capas
Map.add_basemap("Esri.WorldImagery")
Map.addLayer(compuesto, {"bands": ["B4", "B3", "B2"], "min": 0, "max": 0.3}, "Sentinel-2 RGB (mediana ene-mar 2024)")
Map.addLayer(compuesto, {"bands": ["B8", "B4", "B3"], "min": 0, "max": 0.4}, "Falso color (NIR-R-G)", False)
Map.addLayer(compuesto.select("NDVI"), {"min": -0.1, "max": 0.9, "palette": ["brown", "yellow", "green", "darkgreen"]}, "NDVI")
Map.addLayer(ee.Terrain.hillshade(srtm).clip(aoi), {"min": 0, "max": 255}, "Sombreado", False)
Map.addLayerControl()
Map
```

> El mapa es interactivo (capas, zoom, inspector). En Colab se renderiza dentro del cuaderno; si no aparece, ejecute
> `geemap.update_package()` y reinicie el entorno. Cada tesela se calcula **en el momento** en los servidores de EE.

# Bloque 5. Reducers: estadísticas por región, estadísticas zonales y series de tiempo

Un **reducer** agrega píxeles: por región (`reduceRegion`), por muchas regiones (`reduceRegions`), por vecindad o a lo
largo del tiempo. Calculamos el NDVI medio por municipio del área metropolitana usando los límites del dataset FAO GAUL
(nivel 2) y una serie temporal de NDVI mensual para el AOI.

```python
# Límites municipales: FAO GAUL 2025 nivel 2 (disponible en el catálogo de EE)
gaul = ee.FeatureCollection("FAO/GAUL/2025/level2").filter(ee.Filter.eq("GAUL1_NAME", "Santander"))
municipios = gaul.filter(ee.Filter.inList("GAUL2_NAME", ["Bucaramanga", "Floridablanca", "Girón", "Piedecuesta", "Lebrija", "Rionegro"]))

zonal = compuesto.select(["NDVI", "NDBI"]).reduceRegions(collection=municipios, reducer=ee.Reducer.mean(), scale=30)
df_zonal = pd.DataFrame([{"municipio": f["properties"]["GAUL2_NAME"], "NDVI_medio": f["properties"].get("NDVI"), "NDBI_medio": f["properties"].get("NDBI")}
                         for f in zonal.getInfo()["features"]]).sort_values("NDVI_medio", ascending=False)
df_zonal.round(3)
```

```python
# Serie de tiempo: NDVI mensual 2019-2024 en el AOI (todo el map/reduce ocurre en el servidor)
def ndvi_mensual(anio_mes):
    anio_mes = ee.List(anio_mes); inicio = ee.Date.fromYMD(anio_mes.get(0), anio_mes.get(1), 1)
    img = (ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED").filterBounds(aoi).filterDate(inicio, inicio.advance(1, "month"))
             .filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 40)).map(enmascarar_nubes).median())
    ndvi = img.normalizedDifference(["B8", "B4"]).reduceRegion(ee.Reducer.mean(), aoi, 100, maxPixels=1e9).get("nd")
    return ee.Feature(None, {"fecha": inicio.format("YYYY-MM"), "ndvi": ndvi})

meses = ee.List([[a, m] for a in range(2019, 2025) for m in range(1, 13)])
serie = ee.FeatureCollection(meses.map(ndvi_mensual))
t = time.time()
df_serie = pd.DataFrame([f["properties"] for f in serie.getInfo()["features"]]).dropna()
print(f"{len(df_serie)} meses en {time.time()-t:.1f} s")
df_serie["fecha"] = pd.to_datetime(df_serie["fecha"])
plt.figure(figsize=(11, 3.5)); plt.plot(df_serie["fecha"], df_serie["ndvi"], "o-", ms=3)
plt.ylabel("NDVI medio del AOI"); plt.title("Serie mensual de NDVI (Sentinel-2) – Bucaramanga AM"); plt.grid(alpha=.3); plt.show()
```

> Observe la **estacionalidad** (dos picos de verdor tras las temporadas de lluvia bimodales) y los meses sin dato por
> nubosidad. En el cuaderno 8.2 modelaremos estas series para detectar anomalías y cambios.

# Bloque 6. Earth Engine como fuente de datos para machine learning

Este es el puente con las clases anteriores. Para entrenar un clasificador necesitamos **muestras etiquetadas**:
`X` = bandas e índices del compuesto + terreno; `y` = clase de cobertura. Como referencia usamos **Dynamic World**
(Google/WRI, 10 m, casi en tiempo real; banda `label` con 9 clases) agregado por moda en el mismo periodo.
Muestreamos puntos estratificados por clase y los traemos como tabla (o los exportamos si son muchos).

```python
CLASES_DW = ["agua", "arboles", "pastizal", "vegetacion_inundada", "cultivos", "arbustos", "construido", "suelo_desnudo", "nieve_hielo"]
dw = (ee.ImageCollection("GOOGLE/DYNAMICWORLD/V1").filterBounds(aoi).filterDate("2024-01-01", "2024-03-31")
        .select("label").reduce(ee.Reducer.mode()).rename("clase").clip(aoi))

# Imagen de variables: bandas S2, índices, elevación y pendiente (todo alineado a 10 m por EE)
variables = (compuesto.select(["B2", "B3", "B4", "B8", "B11", "B12", "NDVI", "NDWI", "NDBI"])
             .addBands(srtm.rename("elevacion")).addBands(pendiente.rename("pendiente")))
apilado = variables.addBands(dw.toInt())

t = time.time()
muestras = apilado.stratifiedSample(numPoints=400, classBand="clase", region=aoi, scale=10, seed=42, geometries=True)
n = muestras.size().getInfo()
df_m = fc_a_df(muestras)                      # ≤ 5.000 filas por llamada; para más, exporte (Bloque 7)
print(f"{n} muestras estratificadas en {time.time()-t:.1f} s")
df_m["clase_nombre"] = df_m["clase"].map(dict(enumerate(CLASES_DW)))
print(df_m["clase_nombre"].value_counts().to_string())
df_m.head()
```

```python
# Las muestras ya sirven para scikit-learn (Clase 4) o Spark (Clase 6). Un vistazo rápido con Random Forest:
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, cohen_kappa_score
X_cols = ["B2", "B3", "B4", "B8", "B11", "B12", "NDVI", "NDWI", "NDBI", "elevacion", "pendiente"]
X, y = df_m[X_cols], df_m["clase_nombre"]
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
rf = RandomForestClassifier(n_estimators=300, random_state=42, n_jobs=-1).fit(Xtr, ytr)
print(f"Exactitud {accuracy_score(yte, rf.predict(Xte)):.3f} | Kappa {cohen_kappa_score(yte, rf.predict(Xte)):.3f} (partición aleatoria: optimista, Clase 6)")
os.makedirs("salidas_clase7", exist_ok=True)
df_m.to_csv("salidas_clase7/muestras_dw_bucaramanga.csv", index=False)
```
